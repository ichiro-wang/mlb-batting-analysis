## Isolation Forest

For the first outlier detection method, we will try an **Isolation Forest**. Our dataset is highly dimensional, and Isolation Forests work well with such data.

```python
anomaly_mask = isolation_forest(data)
```

### Checking Outliers

Taking a look at who our outliers are and how well (or poor) they performed

```python
anomalies_data = reference_data.iloc[anomaly_mask]
anomalies_data = anomalies_data.sort_values("wRC+", ascending=False)

print(anomalies_data)
```
